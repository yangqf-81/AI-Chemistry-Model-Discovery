```python
from pathlib import Path
import sys, json, csv
root = Path.cwd().resolve()
while not (root / 'src' / 'historical_case.py').is_file():
    if root.parent == root:
        raise RuntimeError('请从教学资源库或notebooks目录运行')
    root = root.parent
sys.path.insert(0, str(root / 'src'))
```

```python
from rdkit import Chem
with (root / 'data/internal/train.csv').open(encoding='utf-8-sig', newline='') as f:
    records = list(csv.DictReader(f))
print({'records': len(records), 'positive': sum(int(r['label']) for r in records)})
structure = Chem.MolFromSmiles(records[0]['canonical_smiles'])
assert structure is not None
print({'example_record': records[0]['record_id'], 'atoms': structure.GetNumAtoms(), 'bonds': structure.GetNumBonds()})
```
